```python
import numpy as np
import warnings
import random
import time
import csv
from imvc.cluster import DAIMC
from imvc.datasets import LoadDataset
from imvc.ampute import Amputer
from sklearn.metrics.cluster import adjusted_mutual_info_score, adjusted_rand_score
from scipy.stats import pearsonr
from sklearn.metrics import mean_squared_error

warnings.filterwarnings("ignore", category=RuntimeWarning)
warnings.filterwarnings("ignore", category=FutureWarning)

def algorithm_testing(nb_cluster, datasets, true_values_datasets, nb_run, datasets_name, algorithm_name, p=0, mechanism="X", missing_values: bool=False):
    AMI_python_values = []
    ARI_python_values = []
    AMI_matlab_values = []
    ARI_matlab_values = []
    python_execution_time = 0
    matlab_execution_time = 0
    
    # Write file header : dataset_name, algo, nb_run_total, current_run, engine, missing_values, amputer_mechanism, AMI, ARI, execution_time
    # with open("test_results.csv", "a", newline='') as csv_file:
    #     write = csv.writer(csv_file, delimiter=';')
    #     write.writerow(["datasets_name", "algorithm_name", "nb_run", "current_run", "engine", "missing_values_percentage", "amputer_mechanism", "AMI", "ARI", "execution_time"])


    for i in range(nb_run):
        random_state = random.randint(0, 2**32 - 1)

        # Add missing values
        if(missing_values):
            transformer = Amputer(p=p, mechanism=mechanism, random_state=random_state)
            datasets = transformer.fit_transform(datasets)

        # Python Engine
        start = time.time()
        evaluate_python = DAIMC(n_clusters=nb_cluster, random_state=random_state, engine="python")
        python_label = evaluate_python.fit_predict(datasets)
        end = time.time()
        python_elasped_time = end - start
        python_execution_time += end - start
        
        
        
        # MATLAB Engine
        start = time.time()
        evaluate_matlab = DAIMC(n_clusters=nb_cluster, random_state=random_state, engine="matlab")
        matlab_labels = evaluate_matlab.fit_predict(datasets)
        end = time.time()
        matlab_elasped_time = end - start
        matlab_execution_time += end - start
        ##
        # Ecrire résultat du run matlab dans fichier
        ##

        # Indicators
        ## Python
        AMI_python_values.append(adjusted_mutual_info_score(python_label, true_values_datasets))
        ARI_python_values.append(adjusted_rand_score(python_label, true_values_datasets))

        ## MATLAB
        AMI_matlab_values.append(adjusted_mutual_info_score(matlab_labels, true_values_datasets))
        ARI_matlab_values.append(adjusted_rand_score(matlab_labels, true_values_datasets))

    return np.mean(AMI_matlab_values), np.mean(AMI_python_values), np.mean(ARI_matlab_values), np.mean(ARI_python_values)

Xs, true_values = LoadDataset.load_dataset(dataset_name="nutrimouse", return_y=True)
algorithm_testing(5, Xs, true_values["diet"], 2, "nutrimouse", "DAIMC")        
```
